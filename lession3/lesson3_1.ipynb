```python
# 讓使用者輸入梯形的上底、下底及高，程式會計算梯形的面積(上底加下底乘以高除以2)
top = float(input("請輸入梯形的上底(公分):"))
bottom = float(input("請輸入梯形的下底(公分):"))
height = float(input("請輸入梯形的高(公分):"))
area = (top + bottom) * height / 2
print("梯形的面積:", area, "平方公分")
```

```python
#!usr/bin/python3
'''
讓使用者輸入圓柱底面的半徑與高，程式會計算圓柱的體積(圓周率乘以半徑乘以半徑乘以高)
'''
# 步驟 1:設定圓周率變數 pi,值為 3.14159
pi = 3.14159

# 請完成:讀取使用者輸入的半徑,並轉成數字,存入 radius
# 步驟 2:讀取使用者輸入的半徑(公分),用 float() 轉成數字,存入變數 radius
radius = float(input("請輸入半徑(公分)："))

# 請完成:讀取使用者輸入的高,並轉成數字,存入 height
# 步驟 3:讀取使用者輸入的高(公分),用 float() 轉成數字,存入變數 height
height = float(input("請輸入高(公分)："))

# 請完成:用公式 pi * radius * radius * height 計算體積,存入 volume
# 步驟 4:用公式「pi × 半徑 × 半徑 × 高」計算體積,存入變數 volume
volume = pi * radius * radius * height

# 步驟 5:用 print() 印出體積,並加上「立方公分」單位
print('圓柱的體積:', volume, '立方公分')
```

```python
"""
讓使用者輸入直角三角形的對邊與斜邊,
利用 sin(x) = 對邊 / 斜邊 計算夾角,
再把弧度(radian)轉換成角度(degree)印出。
"""

import math

opposite = float(input('請輸入直角三角形的對邊長度:'))
hypotenuse = float(input('請輸入直角三角形的斜邊長度:'))
radian = math.asin(opposite / hypotenuse)
degree = math.degrees(radian)
print('夾角的弧度:', radian, '弧度')
print('夾角的角度:', degree, '度')
```
