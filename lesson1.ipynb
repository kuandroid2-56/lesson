### 計算直角三角形的斜邊
根據畢氏定理：$c = \sqrt{a^2 + b^2}$
當兩邊分別為 3 和 4 時：

```python
import math

a = 3
b = 4
c = math.sqrt(a**2 + b**2)

print(f"當兩邊為 {a} 和 {b} 時，斜邊為: {c}")
```

```python
import statistics

data = [3, 6, 9, 11, 17, 25, 32]
stdev = statistics.stdev(data)

print(f"數據集: {data}")
print(f"標準差 (樣本): {stdev}")
```
